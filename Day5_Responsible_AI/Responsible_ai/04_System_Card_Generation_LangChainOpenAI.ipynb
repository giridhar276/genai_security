# Model / System Card

**Use case:** Generate a system card for a high-impact AI use case.

This notebook demonstrates the Day 4 Responsible AI / Governance concept in a simple end-to-end flow.

**Total steps:** 10

## Installation

```bash
pip install pandas numpy scikit-learn matplotlib langchain-openai python-dotenv
```

## LLM setup
Create a `.env` file:

```text
OPENAI_API_KEY=your_key_here
```

This notebook uses `langchain_openai.ChatOpenAI`.

## Step 1 - Define the AI system

````python
system_info = {'name':'Loan Recommendation Assistant','purpose':'Recommend whether an application should receive manual review','users':'Internal credit operations','data':'Customer financial attributes','model':'Logistic Regression + LLM explanation','impact':'High'}
print(system_info)
````

## Step 2 - Define intended use

````python
intended_use = ['Decision support only','Human approval required','No automatic final loan rejection']
print(intended_use)
````

## Step 3 - Define limitations

````python
limitations = ['Historical data can contain bias','Predictions may be wrong','Not suitable for fully autonomous decisions']
print(limitations)
````

## Step 4 - Define risk controls

````python
controls = ['Bias testing','PII protection','Human review','Audit logging','Periodic monitoring']
print(controls)
````

## Step 5 - Build structured context

````python
context = {'system':system_info,'intended_use':intended_use,'limitations':limitations,'controls':controls}
print(context)
````

## Step 6 - Initialize ChatOpenAI

````python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
load_dotenv()
llm = ChatOpenAI(model='gpt-4o-mini',temperature=0)
````

## Step 7 - Create a system-card prompt

````python
prompt = f'''Create a concise AI system card with sections: Purpose, Intended Use, Data, Model, Risks, Limitations, Human Oversight, Monitoring, and Controls. Use only this information: {context}'''
````

## Step 8 - Generate the system card

````python
response = llm.invoke(prompt)
system_card = response.content
print(system_card)
````

## Step 9 - Add review metadata

````python
review_record = {'owner':'AI Governance Team','review_status':'DRAFT','requires_human_approval':True}
print(review_record)
````

## Step 10 - Save the artifact

````python
from pathlib import Path
Path('loan_system_card.md').write_text(system_card,encoding='utf-8')
print('Saved loan_system_card.md')
````
