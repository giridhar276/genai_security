# Governance Evidence & Audit Trail

**Use case:** Create traceable evidence for an AI decision.

This notebook demonstrates the Responsible AI / Governance concept in a simple end-to-end flow.

**Total steps:** 10

## Installation

```bash
pip install pandas numpy scikit-learn matplotlib langchain-openai python-dotenv
```

## Step 1 - Create an AI decision event

````python
event = {'event_id':'EVT001','timestamp':'2026-08-13T09:30:00','system':'Loan Recommendation Assistant','input_ref':'APP1001','prediction':'REVIEW','confidence':0.82,'human_required':True}
print(event)
````

## Step 2 - Add policy information

````python
event['policy_version'] = 'loan-ai-policy-v1.2'
event['model_version'] = 'loan-model-v3'
````

## Step 3 - Add risk evidence

````python
event['risk_checks'] = {'bias_check':'PASS','pii_check':'PASS','grounding_check':'NOT_APPLICABLE'}
print(event['risk_checks'])
````

## Step 4 - Add human oversight status

````python
event['human_decision'] = 'PENDING'
event['reviewer'] = None
````

## Step 5 - Create a hash of the event

````python
import json,hashlib
serialized = json.dumps(event,sort_keys=True,default=str)
event_hash = hashlib.sha256(serialized.encode()).hexdigest()
print(event_hash)
````

## Step 6 - Create an audit record

````python
audit_record = {'event':event,'sha256':event_hash}
print(audit_record)
````

## Step 7 - Save as JSONL

````python
from pathlib import Path
Path('ai_audit_log.jsonl').write_text(json.dumps(audit_record)+'\n',encoding='utf-8')
print('Saved ai_audit_log.jsonl')
````

## Step 8 - Read the evidence back

````python
loaded = json.loads(Path('ai_audit_log.jsonl').read_text(encoding='utf-8').strip())
print(loaded['event']['event_id'])
````

## Step 9 - Verify integrity

````python
check = json.dumps(loaded['event'],sort_keys=True,default=str)
verified = hashlib.sha256(check.encode()).hexdigest()==loaded['sha256']
print('Integrity verified:',verified)
````

## Step 10 - Define retention metadata

````python
retention = {'artifact':'ai_audit_log.jsonl','retention_period':'per organizational policy','owner':'AI Governance'}
print(retention)
````
