## Step 1 - Load oversight cases

````python
import pandas as pd
from pathlib import Path
df = pd.read_csv(Path('data/human_oversight_cases.csv'))
df
````

## Step 2 - Define high-impact actions

````python
high_impact = ['Approve loan','Refund $1200','Reset password','Reject candidate','Recommend treatment']
print(high_impact)
````

## Step 3 - Create an approval rule

````python
df['approval_required'] = df['proposed_action'].isin(high_impact) | (df['impact']=='High')
print(df[['case_id','proposed_action','approval_required']])
````

## Step 4 - Add confidence-based review

````python
df['low_confidence_review'] = df['confidence'] < 0.90
````

## Step 5 - Calculate final review decision

````python
df['final_review_required'] = df['approval_required'] | df['low_confidence_review']
print(df[['case_id','final_review_required']])
````
