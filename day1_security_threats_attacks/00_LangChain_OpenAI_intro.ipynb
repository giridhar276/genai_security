
## 1. Configure the OpenAI API Key

The OpenAI API key should normally be stored outside your source code.

Create a file named `.env` in the same folder as this notebook:

```text
OPENAI_API_KEY=your_openai_api_key_here
```

The `python-dotenv` library loads the value into the environment.

**Do not share or commit your real API key.**


````python
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")

if api_key:
    print("OPENAI_API_KEY loaded successfully.")
else:
    print("OPENAI_API_KEY is not available. Add it to your .env file.")
````

## 2. Create the LLM Object

`ChatOpenAI` is LangChain's integration for OpenAI chat models.

The basic pattern used throughout this notebook is:

```python
response = llm.invoke("your prompt")
print(response.content)
```

`invoke()` sends one input to the model.

The returned value is an `AIMessage`. The generated text can normally be read from `response.content`.

If the model name below is not enabled for your OpenAI account, replace it with a chat model available to you.


````python
from langchain_openai import ChatOpenAI

MODEL_NAME = "gpt-4.1-mini"

llm = ChatOpenAI(
    model=MODEL_NAME,
    temperature=0
)

print("LLM object created.")
````

## 3. First `llm.invoke()` Example

We start with the smallest possible example.

The prompt is passed directly as a Python string.


````python
response = llm.invoke("What is Generative AI? Explain in 3 simple sentences.")

print(response.content)
````

## 4. Sentiment Analysis

Sentiment analysis identifies the emotional tone of text.

Typical labels are:
- Positive
- Negative
- Neutral

Here we ask the LLM to return only one sentiment label.

````python
customer_review = "The delivery was fast and the product quality is excellent."

prompt = f"""
Identify the sentiment of the following customer review.
Return only one word: Positive, Negative, or Neutral.

Review: {customer_review}
"""

response = llm.invoke(prompt)

print("Review   :", customer_review)
print("Sentiment:", response.content)
````

## 5. Language Translation

An LLM can translate text from one language to another.

In this example, an English sentence is translated into Telugu.

````python
english_text = "Your order has been shipped and will arrive tomorrow."

prompt = f"""
Translate the following English text into Telugu.
Return only the translated sentence.

Text: {english_text}
"""

response = llm.invoke(prompt)

print("English:", english_text)
print("Telugu :", response.content)
````

## 6. Text Summarization

Summarization converts a longer passage into a shorter version while preserving the main meaning.

This is useful for emails, reports, support cases, meeting notes, and articles.

````python
long_text = """
Artificial intelligence is increasingly used by organizations to automate repetitive work,
analyze large amounts of data, assist employees, improve customer service, and support
decision-making. Generative AI extends these capabilities by creating new text, code,
images, summaries, and other forms of content from user instructions.
"""

prompt = f"""
Summarize the following text in two simple sentences.

Text:
{long_text}
"""

response = llm.invoke(prompt)

print(response.content)
````

## 7. Text Classification

Classification assigns text to a predefined category.

In this example, a customer message must be classified into one of four departments.

````python
customer_message = "I was charged twice for the same order."

prompt = f"""
Classify the customer message into exactly one category:

Billing
Delivery
Product
Account

Customer message: {customer_message}

Return only the category name.
"""

response = llm.invoke(prompt)

print("Message :", customer_message)
print("Category:", response.content)
````

## 8. Information Extraction

Information extraction asks the model to find specific facts from unstructured text.

Here we extract a customer name, product, order ID, and issue.

````python
support_text = """
Ravi contacted customer support regarding order ORD-1048.
He purchased a Bluetooth speaker, but the device is not turning on.
"""

prompt = f"""
Extract the following information from the text:

Customer Name:
Order ID:
Product:
Issue:

Text:
{support_text}
"""

response = llm.invoke(prompt)

print(response.content)
````

## 9. Grammar Correction

LLMs can correct grammar and spelling while retaining the original meaning.

This can be useful for emails, reports, chat messages, and documentation.

````python
sentence = "I has completed the report and send it yesterday."

prompt = f"""
Correct the grammar of the following sentence.
Return only the corrected sentence.

Sentence: {sentence}
"""

response = llm.invoke(prompt)

print("Original :", sentence)
print("Corrected:", response.content)
````

## 10. Professional Rewriting

The same information can be rewritten in a more professional style.

This is a common business use case for emails and workplace communication.

````python
message = "send me the file fast because i need it today"

prompt = f"""
Rewrite the following message in a polite and professional business style.
Keep it short.

Message: {message}
"""

response = llm.invoke(prompt)

print("Original:", message)
print("Rewritten:", response.content)
````

## 11. Simple Question Answering

A basic LLM can answer general questions directly from its model knowledge.

This example does **not** use RAG or an external database.

````python
question = "What is the difference between machine learning and deep learning?"

prompt = f"""
Answer the following question for a beginner.
Use no more than 4 simple sentences.

Question: {question}
"""

response = llm.invoke(prompt)

print(response.content)
````

## 12. Keyword Extraction

Keyword extraction identifies the most important words or short phrases in a passage.

This can help with document tagging, search, routing, and content organization.

````python
text = """
The company is implementing generative AI applications with prompt engineering,
retrieval augmented generation, security guardrails, model monitoring, and AI governance.
"""

prompt = f"""
Extract 5 important keywords or short phrases from the following text.
Return them as a comma-separated list.

Text:
{text}
"""

response = llm.invoke(prompt)

print(response.content)
````

## 13. Product Description Generation

Generative AI can create simple marketing or catalog content from structured product details.

The input below contains a few product attributes.

````python
product_name = "SmartFit Watch"
features = "heart-rate monitoring, sleep tracking, 7-day battery, water resistant"

prompt = f"""
Write a simple 3-sentence ecommerce product description.

Product: {product_name}
Features: {features}
"""

response = llm.invoke(prompt)

print(response.content)
````

## 14. Multiple Sentiment Examples Using a Loop

Once participants understand one `invoke()` call, they can repeat it for several inputs.

This example processes a small Python list. It is still intentionally simple.

````python
reviews = [
    "Amazing product. I am very happy with it.",
    "The package arrived three days late.",
    "The product is okay and works as expected."
]

for review in reviews:
    prompt = f"""
    Identify the sentiment as Positive, Negative, or Neutral.
    Return only the sentiment.

    Review: {review}
    """

    response = llm.invoke(prompt)

    print("Review   :", review)
    print("Sentiment:", response.content)
    print("-" * 60)
````

## 15. Key Pattern to Remember

Almost every basic example in this notebook follows the same four steps:

```python
# 1. Prepare the input
text = "some input"

# 2. Prepare the instruction
prompt = f"Do something with this text: {text}"

# 3. Invoke the model
response = llm.invoke(prompt)

# 4. Read the generated content
print(response.content)
```
